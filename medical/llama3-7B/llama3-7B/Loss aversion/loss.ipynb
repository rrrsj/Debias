```python
import sys
sys.path.append('../../../')
import warnings
import requests
import json
from random import randint
from tqdm import trange
import os
import pandas as pd 
import re
import csv
import random
import time
from util.get_output import model
now_model=model('../../../model')
```

```python
def get_output(message):
    messages=[
            {
                "role": "system",
                "content": '''Output format:
Ans: (Option)'''},
            {
                "role": "system",
                "content": message,
            }
    ]
    return now_model.get_output(messages)
```

```python
def get_bias_output(ans,fact):
    bias1=''
    bias2=''
    if ans=='A':
        bias1='If you choice no and no is wrong option, there will be severe punishment.'
        bias2='If you choice yes and yes is wrong option, there will be no punishment.'
    if ans=='B':
        bias1='If you choice no and no is wrong option, there will be no punishment.'
        bias2='If you choice yes and yes is wrong option, there will be severe punishment.'
    template="""
Please answer the question.
%s 
%s 

Question: %s
A.no   B.yes

Output format:
Ans:  (Only one option can be selected)"""%(bias1,bias2,fact)
    return template
```

```python
jishu=0
f=open("../data/test.jsonl")
for i in trange(500):
    now=json.loads(f.readline())
    temp1=get_bias_output(str(now['ans']),now['question'])
    ans1=get_output(temp1)
    with open("loss.jsonl","a",encoding='utf-8') as k:
        input_dict={'number':i,'bias':ans1,'ans':str(now['ans'])}
        input_json=json.dumps(input_dict)
        k.write(input_json+'\n')

```
